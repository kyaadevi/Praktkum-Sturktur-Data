Implementasi Bubble Sort

```python
def bubble_sort(arr):
  n = len(arr)
  for i in range(n - 1):
    for j in range(n - i - 1):
      if arr[j] > arr[j + 1]:
        arr[j], arr[j + 1] = arr[j + 1], arr[j]
  return arr
# Contoh penggunaan
data = [64, 34, 25, 12, 22, 11, 90]
sorted_data = bubble_sort(data.copy()) # Menggunakan copy agar data asli tidak berubah
print("Data sebelum sorting:", data)
print("Data setelah Bubble Sort:", sorted_data)
```

Implementasi Selection Sort

```python
def selection_sort(arr):
  n = len(arr)
  for i in range(n - 1):
    min_index = i
    for j in range(i + 1, n):
      if arr[j] < arr[min_index]:
        min_index = j
    arr[i], arr[min_index] = arr[min_index], arr[i]
  return arr
# Contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = selection_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Selection Sort:", sorted_data)
```

Implementasi Insertion Sort

```python
def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
      key = arr[i]
      j = i -1
      while j >= 0 and arr[j] > key:
        arr[j + 1 ] = arr[j]
        j -= 1
      arr[j + 1] = key
    return arr

#contoh penggunaan
data = [64, 34, 12, 22, 11, 90]
sorted_data = insertion_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah sorting:", sorted_data)
```

Implementasi Merge Sort

```python
def merge_sort(arr):
  if len(arr) > 1:
    mid = len(arr) //2
    left_half = arr[:mid]
    right_half = arr[mid:]

    merge_sort(left_half)
    merge_sort(right_half)

    i = j = k = 0

    while i < len(left_half) and j < len(right_half):
      if left_half[i] < right_half[j]:
        arr[k] = left_half[i]
        i += 1
      else:
        arr[k] = right_half[j]
        j += 1
        k += 1

    while i < len(left_half):
      arr[k] = left_half[i]
      i += 1
      k += 1

    while j < len(right_half):
      arr[k] = right_half[j]
      j += 1
      k += 1
  return arr

#contoh penggunaan
data = [64, 25, 12, 22, 11]
sorted_data = merge_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah Merge Sort:", sorted_data)
```

Implementasi Quick Sort

```python
def quick_sort(arr):
  if len(arr) <= 1:
    return arr

  pivot = arr[len(arr) // 2 ]
  left = [x for x in arr if x < pivot]
  middle = [x for x in arr if x == pivot]
  right = [x for x in arr if x > pivot]

  return quick_sort(left) + middle + quick_sort(right)

# Contoh Penggunnaan
data = [64, 25, 12, 22, 11]
sorted_data = quick_sort(data.copy())
print("Data sebelum sorting:", data)
print("Data setelah quick sort:", sorted_data)
```

# Optimasi Pencarian Produk di Marketplace (Rombel 02)
Marketplace sering menampilkan produk berdasarkan harga terendah hingga
tertinggi atau berdasarkan rating pelanggan. Buat program yang mengurutkan
daftar produk menggunakan algoritma sorting.

```python
class Produk:
  def __init__(self, nama, harga, rating):
    self.nama = nama
    self.harga = harga
    self.rating = rating

def quick_sort_produk(arr, key=lambda x:x.harga, ascending=True):
  if len(arr) <= 1:
    return arr


  pivot = arr[len(arr) //2 ]
  pivot_value = key(pivot)

  left = [x for x in arr if key(x) < pivot_value]
  middle = [x for x in arr if key(x) == pivot_value]
  right = [x for x in arr if key(x) > pivot_value]

  if ascending:
    return quick_sort_produk(left, key, ascending) + middle + quick_sort_produk(right, key, ascending)
  else:
    return quick_sort_produk(right, key, ascending) + middle + quick_sort_produk(left, key, ascending)

daftar_produk = [
    Produk("laptop", 12000000, 4.6),
    Produk("mouse", 150000, 4.2),
    Produk("keyboard", 500000, 5.0),
    Produk("charger", 750000, 4.5),
    Produk("monitor", 20000000, 4.9)
]

sorted_by_price = quick_sort_produk(daftar_produk.copy(), key=lambda x: x.harga, ascending=True)
sorted_by_rating = quick_sort_produk(daftar_produk.copy(), key=lambda x: x.rating, ascending=False)


print("Sorting by price (termurah ke termahal):")
for i, p in enumerate(sorted_by_price, 1):
    print(f"{i}. {p.nama} - Rp{p.harga}")

print("\nSorting by rating (tertinggi ke terendah):")
for i, p in enumerate(sorted_by_rating, 1):
    print(f"{i}. {p.nama} - {p.rating}")
```
