Representasi Graf dengan Adjacency List

```python
# Membuat graf tak berarah menggunakan adjacency list
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

# Menampilkan graf
def print_graph(graf):
    for node in graf:
      print(f"(node) -> {graf[node]}")

print_graph(graph)
```

Traversal Graf Menggunakan BFS

```python
from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])

  while queue:
    vertex = queue.popleft()
    visited.add(vertex)
    queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# contoh pemanggilan
bfs(graph, 'A')
```

Traversal Graf menggunakan DFS

```python
def dfs(graph, start, visited=None):
  if visited is None:
    visited = set ()

  if start not in visited:
    print(start, end="")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

# contoh pemanggilan
dfs(graph, 'A')
```

```python
def count_neighbors(graph):
  result = {}
  for node in graph:
    result[node] = len(graph[node])
  return result

graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print(count_neighbors(graph))
```

2. Implementasikan BFS dan DFS dari simpul awal pilihan Anda

```python
from collections import deque

# BSF
def bsf(graph, start):
  visited = set()
  queue = deque([start])
  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      print(vertex, end="")
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])

# DFS
def dfs (graph, start, visited=None):
  if visited is None:
    visited = set()
  if start not in visited:
    print(start, end="")
    visited.add(start)
    for neighbor in graph[start]:
      dfs(graph, neighbor, visited)

print("BFS mulai dari A:")
bsf(new_graph, 'A')
print("\nDFS mulai dari A:")
dfs(new_graph, 'A')
```

3. Modifikasi fungsi BFS agar mengembalikan **list urutan kunjungan**, bujan hanya mencetak

```python
def bfs_return_list(graph, start):
  visited = set()
  queue = deque([start])
  order = []
  while queue:
    vertex = queue.popleft()
    if vertex not in visited:
      order.append(vertex)
      visited.add(vertex)
      queue.extend([neighbor for neighbor in graph[vertex] if neighbor not in visited])
  return order

print(bfs_return_list(new_graph, 'A'))
```

4. Buat fungsi find_path(graph, start, end) yang mengembalikan satu jalur dari start ke end jika ada, menggunakan DFS

```python
def find_path(graph, start, end, path=None):
  if path is None:
    path = []
  path = path + [start]
  if start == end:
    return path
  if start not in graph:
    return None
  for neighbor in graph[start]:
    if neighbor not in path:
      new_path = find_path(graph, neighbor, end, path)
      if new_path:
        return new_path
  return None

print(find_path(new_graph, 'A', 'E'))
```

5. Tambahkan fungsi is_connected(grafh) untuk mengecek apakah semua simpul dalam graf saling terhubung (gunakan kombinasi DFS + logika)

```python
def is_connected(graph):
  if not graph:
    return True
  start = next(iter(graph))
  visited = set()

  def dfs_visit(node):
    visited.add(node)
    for neighbor in graph[node]:
      if neighbor not in visited:
        dfs_visit(neighbor)

  dfs_visit(start)
  return len(visited) == len(graph)

print(is_connected(new_graph))
```
